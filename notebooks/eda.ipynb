```python
# Water Quality EDA Notebook

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Works when launched from repository root or notebooks/
from pathlib import Path
base_path = Path.cwd()
if not (base_path / 'data' / 'water_potability.csv').exists():
    base_path = base_path.parent
data_path = base_path / 'data' / 'water_potability.csv'
df = pd.read_csv(data_path)
```

```python

# Basic Info
print("Shape:", df.shape)
print(df.info())
print(df.describe())
```

```python

# Missing Values
print("\nMissing Values:\n", df.isnull().sum())
```

```python
# EDA only; model preprocessing fits on training data only.
df.fillna(df.median(), inplace=True)
```

```python
# Target distribution
plt.figure()
sns.countplot(x='Potability', data=df)
plt.title('Water Potability Distribution')
plt.show()

```

```python
# Insight
# 0 = Not Safe, 1 = Safe
```

```python
# Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
# Histogram plots
df.hist(figsize=(12,10))
plt.suptitle('Feature Distributions')
plt.show()
```

```python
# Boxplots for outliers
plt.figure(figsize=(12,8))
for i, col in enumerate(df.columns[:-1]):
    plt.subplot(3,3,i+1)
    sns.boxplot(y=df[col])
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
# Potability vs Features
features = df.columns[:-1]

plt.figure(figsize=(15,10))
for i, col in enumerate(features):
    plt.subplot(3,3,i+1)
    sns.boxplot(x='Potability', y=col, data=df)
plt.tight_layout()
plt.show()
```

```python

# Pairplot
sns.pairplot(df, hue='Potability')
plt.show()
```

```python


# Insights
print("""
Key Insights:
1. Dataset had missing values which were filled using median.
2. Data is slightly imbalanced (more unsafe water samples).
3. Some features like Solids, Conductivity show wide variation.
4. Correlation between features is generally low.
5. Outliers exist in multiple columns.
6. Certain features show different distributions for safe vs unsafe water.
""")
```
